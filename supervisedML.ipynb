```python
# ML machine learning supervised usuli orqali kichik task

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

np.random.seed(42)

dars_soati = np.random.uniform(2, 20, 50).reshape(-1, 1)

imtihon_bali = dars_soati.flatten() * 4.5 + np.random.normal(0, 5, 50) + 15

model = LinearRegression()
model.fit(dars_soati, imtihon_bali)

bashorat_ballari = model.predict(dars_soati)

print(f"Koeffitsiyent (1 soat dars necha ball beradi): {model.coef_[0]:.2f}")
print(f"Boshlang'ich ball (0 soat dars qilsa):      {model.intercept_:.2f}")
print(f"Model aniqligi (R2 score):                  {model.score(dars_soati, imtihon_bali):.3f}")

yangi_talaba = np.array([[15]])
yangi_bashorat = model.predict(yangi_talaba)
print(f"\n15 soat dars qilgan talabaning taxminiy bali: {yangi_bashorat[0]:.1f} ball")

```

```python
# Grafik chizish (Vizualizatsiya)
plt.figure(figsize=(8, 5))
plt.scatter(dars_soati, imtihon_bali, color="green", alpha=0.7, label="Haqiqiy talabalar natijasi")
plt.plot(dars_soati, bashorat_ballari, color="red", linewidth=2, label="ML Modelining to'g'ri chizig'i")
plt.xlabel("Haftalik dars qilish vaqti (soat)")
plt.ylabel("Imtihon bali (0-100)")
plt.title("O'qish vaqti va imtihon bali o'rtasidagi bog'liqlik")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
